## 11. Separate Features and Target

```python
X = df.drop(
    columns=["TARGET"]
)

y = df["TARGET"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set :", X_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

print("Number of numerical features:", len(numerical_features))
print("Number of categorical features:", len(categorical_features))
```

```python
print("Categorical features:")
print(categorical_features)
```

## 14. Data Preprocessing

### Numerical features
- Median imputation
- StandardScaler

### Categorical features
- Most-frequent imputation
- One-Hot Encoding

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore"
    ))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python
X_train_processed = preprocessor.fit_transform(
    X_train
)

X_test_processed = preprocessor.transform(
    X_test
)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape :", X_test_processed.shape)
```

## 15. Class Imbalance — Random Oversampling

Oversampling will be applied only to the training data.

The test data remains untouched so that evaluation represents the original
real-world class distribution.

```python
from imblearn.over_sampling import RandomOverSampler
```

```python
print("Before oversampling:")
print(y_train.value_counts())
```

```python
ros = RandomOverSampler(
    random_state=42
)

X_train_resampled, y_train_resampled = ros.fit_resample(
    X_train_processed,
    y_train
)
```

```python
print("After oversampling:")
print(y_train_resampled.value_counts())
```

```python
print("Original training shape:", X_train_processed.shape)
print("Resampled training shape:", X_train_resampled.shape)
```

## 16. Logistic Regression

Logistic Regression will serve as our first classification baseline.

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(
    X_train_resampled,
    y_train_resampled
)

print("Logistic Regression training completed.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
```

```python
y_pred_lr = logistic_model.predict(
    X_test_processed
)

y_prob_lr = logistic_model.predict_proba(
    X_test_processed
)[:, 1]
```

```python
print("Logistic Regression Results")
print("--------------------------------")

print("Accuracy :", round(accuracy_score(y_test, y_pred_lr), 4))
print("Precision:", round(precision_score(y_test, y_pred_lr), 4))
print("Recall   :", round(recall_score(y_test, y_pred_lr), 4))
print("F1 Score :", round(f1_score(y_test, y_pred_lr), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob_lr), 4))
```

```python
print(
    classification_report(
        y_test,
        y_pred_lr
    )
)
```

```python
cm_lr = confusion_matrix(
    y_test,
    y_pred_lr
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

## 20. Artificial Neural Network

We will build an ANN using TensorFlow/Keras and compare its performance
with our classical machine learning models.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

print("TensorFlow version:", tf.__version__)
```

```python
X_train_ann = X_train_resampled
X_test_ann = X_test_processed

print("ANN training shape:", X_train_ann.shape)
print("ANN testing shape :", X_test_ann.shape)
```

```python
# Convert data to float32 for TensorFlow
X_train_ann = X_train_resampled.astype(np.float32)
X_test_ann = X_test_processed.astype(np.float32)

print("ANN training shape:", X_train_ann.shape)
print("ANN testing shape :", X_test_ann.shape)
print("Training data type:", X_train_ann.dtype)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

ann_model = Sequential([
    Dense(128, activation="relu", input_shape=(X_train_ann.shape[1],)),
    Dropout(0.3),

    Dense(64, activation="relu"),
    Dropout(0.3),

    Dense(1, activation="sigmoid")
])

ann_model.summary()
```

```python
import tensorflow as tf

ann_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)
```

```python
history = ann_model.fit(
    X_train_ann,
    y_train_resampled,
    validation_split=0.10,
    epochs=15,
    batch_size=512,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
# Predict probabilities on the untouched test set
ann_prob = ann_model.predict(
    X_test_ann,
    batch_size=1024
).ravel()

# Convert probabilities into class predictions
ann_pred = (ann_prob >= 0.5).astype(int)

print("Predictions generated successfully.")
print("Number of predictions:", len(ann_pred))
```

```python
print("ANN Results")
print("--------------------------------")
print("Accuracy :", round(accuracy_score(y_test, ann_pred), 4))
print("Precision:", round(precision_score(y_test, ann_pred), 4))
print("Recall   :", round(recall_score(y_test, ann_pred), 4))
print("F1 Score :", round(f1_score(y_test, ann_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, ann_prob), 4))
```

```python
model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "ANN"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, ann_pred)
    ],
    "Precision": [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_dt),
        precision_score(y_test, y_pred_rf),
        precision_score(y_test, ann_pred)
    ],
    "Recall": [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_dt),
        recall_score(y_test, y_pred_rf),
        recall_score(y_test, ann_pred)
    ],
    "F1 Score": [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_dt),
        f1_score(y_test, y_pred_rf),
        f1_score(y_test, ann_pred)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test, y_prob_lr),
        roc_auc_score(y_test, y_prob_dt),
        roc_auc_score(y_test, y_prob_rf),
        roc_auc_score(y_test, ann_prob)
    ]
})

model_comparison.round(4)
```

```python
print("ANN Classification Report")
print("--------------------------------")
print(classification_report(y_test, ann_pred))
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    confusion_matrix(y_test, ann_pred),
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Default", "Default"],
    yticklabels=["No Default", "Default"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("ANN Confusion Matrix")
plt.show()
```

```python
thresholds = [0.2, 0.3, 0.4, 0.5]

threshold_results = []

models_probabilities = {
    "Logistic Regression": y_prob_lr,
    "Decision Tree": y_prob_dt,
    "Random Forest": y_prob_rf,
    "ANN": ann_prob
}

for model_name, probabilities in models_probabilities.items():

    for threshold in thresholds:

        predictions = (probabilities >= threshold).astype(int)

        threshold_results.append({
            "Model": model_name,
            "Threshold": threshold,
            "Precision": precision_score(y_test, predictions),
            "Recall": recall_score(y_test, predictions),
            "F1 Score": f1_score(y_test, predictions),
            "ROC-AUC": roc_auc_score(y_test, probabilities)
        })

threshold_df = pd.DataFrame(threshold_results)

threshold_df.round(4)
```

```python
best_threshold = (
    threshold_df
    .sort_values("F1 Score", ascending=False)
    .iloc[0]
)

print("Best threshold based on F1 Score:")
print(best_threshold)
```

```python
from sklearn.metrics import roc_curve

# Calculate ROC curve values
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
fpr_dt, tpr_dt, _ = roc_curve(y_test, y_prob_dt)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)
fpr_ann, tpr_ann, _ = roc_curve(y_test, ann_prob)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr_lr, tpr_lr,
    label=f"Logistic Regression (AUC = {roc_auc_score(y_test, y_prob_lr):.3f})"
)

plt.plot(
    fpr_dt, tpr_dt,
    label=f"Decision Tree (AUC = {roc_auc_score(y_test, y_prob_dt):.3f})"
)

plt.plot(
    fpr_rf, tpr_rf,
    label=f"Random Forest (AUC = {roc_auc_score(y_test, y_prob_rf):.3f})"
)

plt.plot(
    fpr_ann, tpr_ann,
    label=f"ANN (AUC = {roc_auc_score(y_test, ann_prob):.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.show()
```
